```python
# 1st all courses under Thayer School of Engineering (graduate level)
url ="https://engineering.dartmouth.edu/courses/graduate"
response1 = requests.get(url)  # send http request
soup_grad_engineer = BeautifulSoup(response1.text, 'html.parser') # parse html
```

```python
grad_course_engineering = []
for li_tag in soup_grad_engineer.find_all('li', class_ ='accordion-item'): # find all course wrappers
    a_tag = li_tag.find('a', class_='accordion-title')
    #print("a_tag", a_tag)
    if not a_tag:
        continue 

       
    spans = a_tag.find_all('span')
    course_code = spans[0].get_text(strip=True)
    course_name = spans[1].get_text(strip=True)

    href = a_tag.get('href')
    content_id = href[1:] if href and href.startswith("#") else None 
    #print("content id:", content_id)

    # get the related description via id 
    
    #print("content id", content_id)
    description_div = li_tag.find("div", id=content_id)
    #print(description_div)
    #print("description div found:", bool(description_div))
    
    course_desc = ""

    if description_div:
        #print('content div html:\n', description_div.prettify() if description_div else "None")
        #print('found decsiption_div')
        content_block = description_div.find('div', class_ = "columns small-12 medium-8")
        #print('found content_block')
        if content_block:
            #h4_tag = content_block.find("h4", string = lambda x: x and "Description" in x)
            h4_tag = None
            for h in content_block.find_all("h4"):
                if "Description" in h.get_text(strip=True):
                    h4_tag = h
                    break
                
            if h4_tag:
                #print("found <h4> Description</h4>")
                desc_text = ""
                for sibling in h4_tag.next_siblings:  # next_siblings is a generator, not a method, like find-all()
                    if getattr(sibling, 'name', None) == 'h4':
                        break
                    if isinstance(sibling, str):
                        desc_text += sibling.strip()
                    elif hasattr(sibling, 'get_text'):
                        desc_text += sibling.get_text(strip = True)
                course_desc = desc_text.strip()
  
            
    
    grad_course_engineering.append({
        'course_code': course_code,
        "course_name": course_name,
        "course_description": course_desc,
        "course_url": url
        })
        
      
```

```python
from selenium.webdriver.chrome.options import Options
chrome_options = Options()
chrome_options.add_argument('--no-sandbox')
chrome_options.add_argument('--disable-dev-shm-usage')
chrome_options.add_argument('--headless')

```

```python
driver = webdriver.Chrome(options=chrome_options)
url5 = "https://tuck.dartmouth.edu/mba/academic-experience/elective-curriculum/elective-courses"
driver.get(url5)

time.sleep(3)
```

```python
# click the fall tab
plus_buttons = driver.find_elements(By.CSS_SELECTOR, 'li.accordion-item button.accordion-title')
for btn in plus_buttons:
    try:
        driver.execute_script('arguments[0].scrollIntoView(true);', btn)
        time.sleep(0.5)
        btn.click()
        time.sleep(0.5)
    except Exception as e:
        print(f"Error clicking: {e}")
        continue 

html = driver.page_source
driver.quit()

```

```python
soup_tuck_elec_fall = BeautifulSoup(html, 'html.parser')
tuck_courses_elective = []

for item in soup_tuck_elec_fall.find_all('li', class_="accordion-item"):
    h3 = item.select_one('a h3.card-title')
    desc = item.select_one('div.accordion-content p.course-description')
    
    
    #print("h3:", h3)
    #print("desc:", desc)
    #print("-" * 40)

    if h3 and desc:
        course_name = h3.get_text(strip=True)
        course_desc = desc.get_text(strip=True)
        course_code = ""
        #print('course name:', course_name)
        #print('course desc:', course_desc)

        if "Course Code" in course_name or "Description" in course_desc:
           continue
        tuck_courses_elective.append({
                "course_code": course_code,
                "course_name": course_name,
                "course_description": course_desc,
                "course_url": url5 })
```

```python
# Biological Sciences - Graduate
url6 ="https://dartmouth.smartcatalogiq.com/en/current/orc/departments-programs-graduate/biological-sciences/biol-biological-sciences/"
response6 = requests.get(url6)
soup_bio_grad = BeautifulSoup(response6.text, 'html.parser')
```

```python

import re
from urllib.parse import urljoin
bio_grad_courses = []
for a in soup_bio_grad.find_all('a', href=True):
    if '/biol-' in a['href'].lower():
        full_text =a.get_text(strip=True)
        #print("Raw:", repr(full_text))
        #print(a['href'])

        normalized = full_text.replace('\xa0','').replace('Â', '').strip()
        #print('normalized:', repr(normalized)) 
       
        
        #full_text = a_tag.get_text(separator = " ", strip=True)
        #cleaned_title = re.sub(r'^BIOL[\s\u00a0]+[\d]+[\s\u00a0]+', "", full_text)
        #match = re.match(r'^BIOL[\s\u00a0]*(\d+)[\s\u00a0]*(.+)$',normalized)
        match = re.match(r'^BIOL[\s\u00a0]*(\d+(?:\.\d+)?)[\s\u00a0]*(.+)$',normalized)

        if match:
            course_code = match.group(1)
            course_name = match.group(2)
            course_name = re.sub(r'^\.\d{1,2}', '',course_name.strip())
            course_url = urljoin(url6, a['href'])  # automatically remove overlapping part 
            

            try:
                course_response = requests.get(course_url)
                course_soup = BeautifulSoup(course_response.text, 'html.parser')
                desc_div = course_soup.find('div', class_='desc')
                course_description = desc_div.get_text(strip=True) if desc_div else ''
            except Exception as e:
                course_description = f"[Error fetching description; {e}]"

          

            bio_grad_courses.append({
                "course_code": course_code,
                "course_name": course_name,
                "course_description": course_description,
                "course_url": course_url,
            })
```

```python
# Health Administration 
# https://mha.dartmouth.edu/curriculum-schedule/core-courses/
url_med4 ="https://mha.dartmouth.edu/curriculum-schedule/core-courses/"
response_med4 = requests.get(url_med4)
soup_med4 = BeautifulSoup(response_med4.text, 'html.parser')
```

```python
import re
med_course_4 = []
for item in soup_med4.find_all('div', class_ = 'js-accordion-item'):
    title_tag = item.find('span', class_ = 'accordion_trigger_title')
    title = title_tag.get_text(strip=True) if title_tag else None

    # get course description
    description_tag = item.find('span', lang='EN')
    if description_tag:
        course_description = description_tag.get_text(strip=True)
    else: 
        content_div = item.find('div', class_ = 'js-accordion-content')
        p_tag = content_div.find('p') if content_div else None
        course_description = p_tag.get_text(strip=True) if p_tag else None
    
    if title and course_description:
        med_course_4.append({
            "course_code": '',
            "course_name": title,
            "course_description": course_description,
            "course_url": url_med4  
        })

    
```

```python
# Health Care Delivery Science
base_url = "https://dartmouth.smartcatalogiq.com"
index_url = f"{base_url}/en/current/orc/departments-programs-graduate/health-care-delivery-science/hcds-courses/"
response = requests.get(index_url)
soup = BeautifulSoup(response.content, "html.parser") 

med_course_5 = [] 
for li in soup.select("ul.sc-child-item-links li a"): 
    raw_title = li.get_text(strip=True)
    cleaned_title = raw_title.replace("\xa0", " ")
    href = li.get('href')
    #title = li.get_text(strip=True) 
    #href = li.get("href") 
    if href: 
        med_course_5.append({ 
            "course_code": '',
            "course_name": cleaned_title, 
            "course_description": '',
            "course_url": base_url + href 
            })
```

```python
for course in med_course_5:
    resp = requests.get(course["course_url"])
    detail_soup = BeautifulSoup(resp.content, "html.parser")

    desc_tag = detail_soup.find("div", class_="desc")
    if desc_tag:
        course["course_description"] = desc_tag.get_text(strip=True) 
    else:
        p_tag = detail_soup.find("p") 
        course["course_description"] = p_tag.get_text(strip=True) if p_tag else "No description found"
```
