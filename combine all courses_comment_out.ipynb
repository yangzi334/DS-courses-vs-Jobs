```python
# all checking courses and courses descriptions include 4 parts:

# 1st: all courses and courses descriptions under Thayer School of Engineering
# 2nd: all courses and courses descriptions under Tuck School of Business
# 3rd: all courses and courses descriptions under 'schools of arts and sciences' and 'Geisel School of Medicine matched programs', already generated and saved under 'dartmouth_matched_ungergrad_med_department.csv'
# 4th: all courses and courses descriptions under Guarini School of Graduate and Advanced Studies  
```

```python
# dartmouth_engineering_courses.csv
import pandas as pd
df_loaded_engineering = pd.read_csv('dartmouth_engineering_courses.csv', header = None, names = ['course_code', 'course_name', 'course_description', "course_url"])
```

```python
df_loaded_engineering
```

```python
# tuck_school_courses_total.csv
import pandas as pd
df_loaded_tuck = pd.read_csv('tuck_school_courses_total.csv', header = None, names = ['course_code','course_name', 'course_description', 'course_url'])

```

```python
# schools of arts and sciences - 'arts_science_grad_courses.csv'
import pandas as pd
df_loaded_arts_sciences = pd.read_csv('arts_science_grad_courses.csv', header = None, names = ['course_code','course_name', 'course_description', 'course_url'])
```

```python
# Geisel School of Medicine courses - 'med_courses.csv'

import pandas as pd
df_loaded_med_courses= pd.read_csv('med_courses.csv', names = ['course_code','course_name', 'course_description', 'course_url'], header = None)
```

```python
# set all above courses data frame together

df_loaded_courses= pd.concat([df_loaded_engineering, df_loaded_tuck, df_loaded_arts_sciences, df_loaded_med_courses], ignore_index = True)
    
```
